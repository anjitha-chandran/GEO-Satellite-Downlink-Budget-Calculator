```python
import math

print("=" * 45)
print("  SATELLITE LINK BUDGET CALCULATOR")
print("=" * 45)

# ── Step 1: Enter your values ──────────────────
print("\nEnter the following values:\n")

power_w    = float(input("Transmit power (Watts)       : "))
gain_tx    = float(input("Transmit antenna gain (dBi)  : "))
freq_ghz   = float(input("Frequency (GHz)              : "))
dist_km    = float(input("Distance (km)                : "))
gain_rx    = float(input("Receive antenna gain (dBi)   : "))
noise_temp = float(input("System noise temperature (K) : "))
bandwidth  = float(input("Bandwidth (MHz)              : "))
```

```python
# ── Step 2: Calculate ──────────────────────────

# Convert watts to dBW
power_dbw = 10 * math.log10(power_w)

# EIRP = transmit power + transmit antenna gain
eirp = power_dbw + gain_tx

# Free space path loss
fspl = 92.4 + 20 * math.log10(freq_ghz) + 20 * math.log10(dist_km)

# Received signal power
power_rx = eirp - fspl + gain_rx

# Noise power
noise_power = 10 * math.log10(1.38e-23 * noise_temp * bandwidth * 1e6)

# C/N ratio
cn_ratio = power_rx - noise_power
```

```python
# ── Step 3: Print results ──────────────────────

print("\n" + "=" * 45)
print("  RESULTS")
print("=" * 45)
print(f"  Transmit power   : {power_dbw:.2f} dBW")
print(f"  EIRP             : {eirp:.2f} dBW")
print(f"  Path loss        : {fspl:.2f} dB")
print(f"  Received power   : {power_rx:.2f} dBW")
print(f"  Noise power      : {noise_power:.2f} dBW")
print(f"  C/N ratio        : {cn_ratio:.2f} dB")
print("=" * 45)
```

```python
# ── Step 4: Does the link close? ───────────────
if cn_ratio >= 5.5:
    print(f"  LINK CLOSES — C/N is {cn_ratio:.2f} dB")
else:
    print(f"  LINK FAILS  — C/N is only {cn_ratio:.2f} dB")
    print("  Try increasing antenna gain or transmit power.")

print("=" * 45)
```
